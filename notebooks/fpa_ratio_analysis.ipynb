```python
df = pd.read_csv('Financial Statements.csv')
df.columns = [c.strip() for c in df.columns]

df.shape
```

```python
df.head()
```

```python
print("Companies:", sorted(df['Company'].unique()))
print("Years:", sorted(df['Year'].unique()))
print("Categories:", df['Category'].unique())
print("\nRows per company:\n", df.groupby('Company')['Year'].count())
print("\nMissing values:\n", df.isnull().sum())
```

```python
df['Category'] = df['Category'].replace({'BANK': 'Bank'})
df['Category'].value_counts()
```

```python
# Spot-check: recompute Net Profit Margin and compare to the given column
df['Net Profit Margin (calc)'] = (df['Net Income'] / df['Revenue']) * 100

# Show the given vs calculated side by side for a few rows
df[['Company', 'Year', 'Net Profit Margin', 'Net Profit Margin (calc)']].head(10)
```

```python
# ROE = Net Income / Shareholder Equity
df['ROE (calc)'] = (df['Net Income'] / df['Share Holder Equity']) * 100

df[['Company', 'Year', 'ROE', 'ROE (calc)']].head(10)
```

```python
df_sorted = df.sort_values(['Company', 'Year'])

df_sorted['Revenue YoY %'] = df_sorted.groupby('Company')['Revenue'].pct_change() * 100
df_sorted['Net Income YoY %'] = df_sorted.groupby('Company')['Net Income'].pct_change() * 100

df_sorted[['Company', 'Year', 'Category', 'Revenue', 'Revenue YoY %', 'Net Income', 'Net Income YoY %']].head(20)
```

```python
sector_trend = df_sorted.groupby(['Category', 'Year'])[['Revenue YoY %', 'Net Income YoY %']].mean().reset_index()
sector_trend
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

def backtest_forecast(company, train_end_year, test_years, metric='Revenue'):
    comp_data = df_sorted[df_sorted['Company'] == company].sort_values('Year')
    train = comp_data[comp_data['Year'] <= train_end_year]
    test = comp_data[comp_data['Year'].isin(test_years)]
    
    X_train = train['Year'].values.reshape(-1, 1)
    y_train = train[metric].values
    
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    X_test = test['Year'].values.reshape(-1, 1)
    predicted = model.predict(X_test)
    actual = test[metric].values
    
    result = test[['Year']].copy()
    result['Actual'] = actual
    result['Predicted'] = predicted
    result['Error %'] = ((predicted - actual) / actual) * 100
    return result

msft_forecast = backtest_forecast('MSFT', train_end_year=2021, test_years=[2022, 2023], metric='Revenue')
msft_forecast
```

```python
nvda_forecast = backtest_forecast('NVDA', train_end_year=2021, test_years=[2022, 2023], metric='Revenue')
nvda_forecast
```

```python
mcd_forecast = backtest_forecast('MCD', train_end_year=2021, test_years=[2022, 2023], metric='Revenue')
mcd_forecast
```

```python
df_sorted.to_csv('financial_statements_cleaned.csv', index=False)
```

```python
backtest_results = pd.concat([
    mcd_forecast.assign(Company='MCD'),
    msft_forecast.assign(Company='MSFT'),
    nvda_forecast.assign(Company='NVDA')
])
backtest_results.to_csv('backtest_results.csv', index=False)
```
